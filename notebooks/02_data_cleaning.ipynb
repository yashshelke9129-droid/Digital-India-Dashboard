```python
import pandas as pd

print("Pandas Version:", pd.__version__)
```

```python
file_path = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\raw\digital_india.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nShape:", df.shape)
```

```python
df.info()
```

```python
print("Missing Values")

df.isnull().sum()
```

```python
total_missing = df.isnull().sum().sum()

print("Total Missing Values:", total_missing)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
df = df.drop_duplicates()

print("Duplicates Removed Successfully")
```

```python
print("Remaining Duplicates:", df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
df["Year"] = df["Year"].astype(int)

df["GDP_Billion"] = df["GDP_Billion"].astype(float)

df["Internet_Users"] = df["Internet_Users"].astype(float)

df["UPI_Billion"] = df["UPI_Billion"].astype(float)

df["Smartphone_Million"] = df["Smartphone_Million"].astype(float)

df["Literacy_Rate"] = df["Literacy_Rate"].astype(float)

print("Data Types Converted Successfully")
```

```python
df.dtypes
```

```python
df.describe()
```

```python
numeric_columns = [
    "GDP_Billion",
    "Internet_Users",
    "UPI_Billion",
    "Smartphone_Million",
    "Literacy_Rate"
]

for col in numeric_columns:
    print("\n", "="*50)
    print(col)
    print("="*50)

    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
```

```python
cleaning_report = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

cleaning_report
```

```python
report_path = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\processed\cleaning_report.csv"

cleaning_report.to_csv(
    report_path,
    index=False
)

print("Cleaning Report Saved Successfully!")
```

```python
cleaned_path = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\processed\digital_india_cleaned.csv"

df.to_csv(
    cleaned_path,
    index=False
)

print("Cleaned Dataset Saved Successfully!")
```
