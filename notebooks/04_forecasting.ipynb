```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression

import matplotlib.pyplot as plt
```

```python
file_path = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\processed\digital_india_cleaned.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
X = df[["Year"]]

y = df["Internet_Users"]

model = LinearRegression()

model.fit(X, y)

print("Model Trained Successfully")
```

```python
future_years = pd.DataFrame({
    "Year":[2026,2027,2028,2029,2030]
})

future_years
```

```python
future_years["Predicted_Internet"] = model.predict(
    future_years
)

future_years
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df["Year"],
    df["Internet_Users"],
    marker="o",
    label="Actual"
)

plt.plot(
    future_years["Year"],
    future_years["Predicted_Internet"],
    marker="o",
    linestyle="--",
    label="Forecast"
)

plt.legend()

plt.title(
    "Internet Users Forecast till 2030"
)

plt.show()
```

```python
forecast_path = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\processed\internet_forecast.csv"

future_years.to_csv(
    forecast_path,
    index=False
)

print("Forecast Saved Successfully")
```
