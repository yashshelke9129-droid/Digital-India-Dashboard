```python
print("Minimum GDP:", df["GDP_Billion"].min())

print("Maximum GDP:", df["GDP_Billion"].max())

print("Average GDP:", round(df["GDP_Billion"].mean(), 2))
```

```python
print("Minimum Internet Users %:", df["Internet_Users"].min())

print("Maximum Internet Users %:", df["Internet_Users"].max())

print("Average Internet Users %:", round(df["Internet_Users"].mean(), 2))
```

```python
print("Minimum Smartphone Users:", df["Smartphone_Million"].min())

print("Maximum Smartphone Users:", df["Smartphone_Million"].max())

print("Average Smartphone Users:", round(df["Smartphone_Million"].mean(), 2))
```

```python
print("Minimum UPI Transactions:", df["UPI_Billion"].min())

print("Maximum UPI Transactions:", df["UPI_Billion"].max())

print("Average UPI Transactions:", round(df["UPI_Billion"].mean(), 2))
```

```python
print("Minimum Literacy Rate:", df["Literacy_Rate"].min())

print("Maximum Literacy Rate:", df["Literacy_Rate"].max())

print("Average Literacy Rate:", round(df["Literacy_Rate"].mean(), 2))
```

```python
summary = pd.DataFrame({
    "Metric": [
        "GDP",
        "Internet Users",
        "UPI Transactions",
        "Smartphone Users",
        "Literacy Rate"
    ],
    "Minimum": [
        df["GDP_Billion"].min(),
        df["Internet_Users"].min(),
        df["UPI_Billion"].min(),
        df["Smartphone_Million"].min(),
        df["Literacy_Rate"].min()
    ],
    "Maximum": [
        df["GDP_Billion"].max(),
        df["Internet_Users"].max(),
        df["UPI_Billion"].max(),
        df["Smartphone_Million"].max(),
        df["Literacy_Rate"].max()
    ]
})

summary
```

```python
report_path = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\processed\collection_report.csv"

summary.to_csv(report_path, index=False)

print("Collection Report Saved Successfully!")
print(report_path)
```
