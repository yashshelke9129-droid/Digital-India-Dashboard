```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

print("Libraries Imported Successfully")
```

```python
file_path = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\processed\digital_india_cleaned.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

df.describe()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df["Year"],
    df["Internet_Users"],
    marker="o",
    linewidth=3
)

plt.title("Internet Users Growth in India (2016-2025)")
plt.xlabel("Year")
plt.ylabel("Internet Users (%)")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df["Year"],
    df["Smartphone_Million"],
    marker="o",
    linewidth=3
)

plt.title("Smartphone Adoption in India")
plt.xlabel("Year")
plt.ylabel("Users (Million)")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df["Year"],
    df["Literacy_Rate"],
    marker="o",
    linewidth=3
)

plt.title("Literacy Rate Growth")
plt.xlabel("Year")
plt.ylabel("Literacy Rate (%)")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=df["Internet_Users"],
    y=df["Smartphone_Million"]
)

plt.title("Internet Users vs Smartphone Adoption")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=df["GDP_Billion"],
    y=df["Internet_Users"]
)

plt.title("GDP vs Internet Users")

plt.show()
```

```python
growth_df = df.copy()

growth_df["GDP_Growth_%"] = (
    growth_df["GDP_Billion"].pct_change() * 100
)

growth_df["Internet_Growth_%"] = (
    growth_df["Internet_Users"].pct_change() * 100
)

growth_df["UPI_Growth_%"] = (
    growth_df["UPI_Billion"].pct_change() * 100
)

growth_df
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    growth_df["Year"],
    growth_df["GDP_Growth_%"],
    marker="o"
)

plt.title("Yearly GDP Growth Rate")

plt.ylabel("Growth %")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    growth_df["Year"],
    growth_df["Internet_Growth_%"],
    marker="o"
)

plt.title("Internet Users Growth Rate")

plt.ylabel("Growth %")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    growth_df["Year"],
    growth_df["UPI_Growth_%"],
    marker="o"
)

plt.title("UPI Growth Rate")

plt.ylabel("Growth %")

plt.grid(True)

plt.show()
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Internet Users",
        "GDP",
        "UPI",
        "Smartphone Users",
        "Literacy Rate"
    ],
    "2016": [
        df["Internet_Users"].iloc[0],
        df["GDP_Billion"].iloc[0],
        df["UPI_Billion"].iloc[0],
        df["Smartphone_Million"].iloc[0],
        df["Literacy_Rate"].iloc[0]
    ],
    "2025": [
        df["Internet_Users"].iloc[-1],
        df["GDP_Billion"].iloc[-1],
        df["UPI_Billion"].iloc[-1],
        df["Smartphone_Million"].iloc[-1],
        df["Literacy_Rate"].iloc[-1]
    ]
})

comparison
```

```python
comparison["Growth_%"] = (
    (comparison["2025"] - comparison["2016"])
    /
    comparison["2016"]
) * 100

comparison
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    comparison["Metric"],
    comparison["Growth_%"]
)

plt.title("Growth Percentage (2016-2025)")
plt.ylabel("Growth %")

plt.xticks(rotation=20)

plt.show()
```

```python
analysis_output = r"C:\Users\Yash 8939\OneDrive\Desktop\Digital-India-Dashboard\data\processed\analysis_summary.csv"

comparison.to_csv(
    analysis_output,
    index=False
)

print("Analysis Summary Saved Successfully")
```
